### 3.1 Measures of Central Tendency — Mean, Median, and Mode

```python
import pandas as pd

numerical_measures = ["Area", "Production"]

central_tendency = pd.DataFrame({
    "Mean": df[numerical_measures].mean(),
    "Median": df[numerical_measures].median(),
    "Mode": [
        df["Area"].mode(dropna=True).iloc[0],
        df["Production"].mode(dropna=True).iloc[0]
    ]
})

display(central_tendency)
```

**Interpretation:**

- **Mean** is the arithmetic average and can be influenced by very large values.
- **Median** is the middle value and is more resistant to extreme observations.
- **Mode** is the most frequently occurring value.

If the mean is much larger than the median, the distribution may be strongly right-skewed.

### 3.2 Measures of Dispersion

```python
q1 = df[numerical_measures].quantile(0.25)
q3 = df[numerical_measures].quantile(0.75)

dispersion = pd.DataFrame({
    "Minimum": df[numerical_measures].min(),
    "Maximum": df[numerical_measures].max(),
    "Range": df[numerical_measures].max() - df[numerical_measures].min(),
    "Variance": df[numerical_measures].var(),
    "Standard Deviation": df[numerical_measures].std(),
    "IQR": q3 - q1
})

display(dispersion)
```

**Interpretation:**

- **Range = Maximum − Minimum**
- **Variance** measures spread around the mean in squared units.
- **Standard deviation** shows spread in the original unit.
- **IQR = Q3 − Q1** measures the spread of the middle 50% of observations.

### 3.3 Measures of Position — Quartiles and Percentiles

```python
position_statistics = pd.DataFrame({
    "Q1 / 25th Percentile": df[numerical_measures].quantile(0.25),
    "Q2 / Median / 50th Percentile": df[numerical_measures].quantile(0.50),
    "Q3 / 75th Percentile": df[numerical_measures].quantile(0.75),
    "90th Percentile": df[numerical_measures].quantile(0.90),
    "95th Percentile": df[numerical_measures].quantile(0.95)
})

display(position_statistics)
```

## 10. Outlier Detection (IQR Method)

### 10.1 Area

```python
Q1_area = df['Area'].quantile(0.25)
Q3_area = df['Area'].quantile(0.75)
IQR_area = Q3_area - Q1_area

lower_area = Q1_area - 1.5 * IQR_area
upper_area = Q3_area + 1.5 * IQR_area

area_outliers = df[(df['Area'] < lower_area) | (df['Area'] > upper_area)]

print("Area Q1:", Q1_area)
print("Area Q3:", Q3_area)
print("Area IQR:", IQR_area)
print("Lower Bound:", lower_area)
print("Upper Bound:", upper_area)
print("Potential Area Outliers:", len(area_outliers))
```

**What is done:** The Interquartile Range (IQR) method is applied to `Area`. Q1 (25th percentile) and Q3 (75th percentile) are computed, the IQR is calculated as Q3 - Q1, and outlier bounds are set at Q1 - 1.5*IQR (lower) and Q3 + 1.5*IQR (upper). Values outside these bounds are flagged as potential outliers.

**Why it is done:** The IQR method is a standard statistical technique for detecting outliers that is robust to extreme values (unlike mean-based methods). It identifies data points that fall significantly outside the central 50% of the distribution.

**What it means for the dataset:** The number of potential area outliers indicates how many records have unusually large or small cultivated areas. These outliers are retained because large agricultural areas are legitimate (e.g., major farming states). The bounds and IQR provide numerical context for the visual outliers seen in the box plot.

### 10.2 Production

```python
Q1_production = df['Production'].quantile(0.25)
Q3_production = df['Production'].quantile(0.75)
IQR_production = Q3_production - Q1_production

lower_production = Q1_production - 1.5 * IQR_production
upper_production = Q3_production + 1.5 * IQR_production

production_outliers = df[(df['Production'] < lower_production) | (df['Production'] > upper_production)]

print("Production Q1:", Q1_production)
print("Production Q3:", Q3_production)
print("Production IQR:", IQR_production)
print("Lower Bound:", lower_production)
print("Upper Bound:", upper_production)
print("Potential Production Outliers:", len(production_outliers))
```

**What is done:** The IQR method is applied to `Production` using the same approach as for `Area`: computing Q1, Q3, IQR, lower and upper bounds, and counting values outside these bounds.

**Why it is done:** Production outliers are especially important to identify since `Production` is the primary production measure. Understanding the outlier structure helps interpret model evaluation metrics — if the model struggles with extreme values, it is likely due to these outliers.

**What it means for the dataset:** Both `Area` and `Production` outliers are **retained**. High cultivated areas and high production values can be legitimate, particularly for highly productive crops or regions — the outliers are not automatically treated as errors. The large number of production outliers explains the right skew observed in earlier visualizations.

### 12.3 Pearson and Spearman Correlation — Area vs Production

```python
pearson_area_production = df[["Area", "Production"]].corr(
    method="pearson"
).loc["Area", "Production"]

spearman_area_production = df[["Area", "Production"]].corr(
    method="spearman"
).loc["Area", "Production"]

print(f"Pearson correlation : {pearson_area_production:.4f}")
print(f"Spearman correlation: {spearman_area_production:.4f}")
```

- **Spearman** summarizes monotonic association using ranks and is less sensitive to extreme values.

Correlation describes association only; it does not prove causation.
